## Part 4: Newton’s Method (10 pts)
### 4.1
Use Newton’s method to approximate $\sqrt{6} $:
- Use $ f(x) = x^2 - 6 $, start at $ x_0 = 6 $
- Perform 5 iterations and print values and errors.

````python
# 4-1 Newton's Method for f(x) = x^2 - 6
def f(x): return x**2 - 6
def df(x): return 2 * x

x_vals = [2.0]
for _ in range(3):
    x_next = x_vals[-1] - f(x_vals[-1]) / df(x_vals[-1])
    x_vals.append(x_next)

for i, x in enumerate(x_vals):
    print(f"x_{i} = {x:.5f}, f(x_{i}) = {f(x):.5e}")
````

### 4.2
Plot $f(x) $ and the tangent lines at each Newton step. Mark and annotate all steps.

````python
#4-2 plot f(x)
import matplotlib.pyplot as plt
import numpy as np

x_plot = np.linspace(0, 3, 400)
y_plot = f(x_plot)

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(x_plot, y_plot, label='f(x) = x^2 - 6', color='blue')
ax.axhline(0, color='black')

for i in range(4):
    x_i = x_vals[i]
    y_i = f(x_i)
    slope = df(x_i)
    tangent_x = np.linspace(x_i - 1, x_i + 1, 100)
    tangent_y = slope * (tangent_x - x_i) + y_i
    ax.plot(tangent_x, tangent_y, '--', color='gray')
    ax.plot(x_i, y_i, 'ro')
    ax.annotate(f"x_{i}", (x_i + 0.05, y_i + 0.5))

ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.set_title("Newton's Method for f(x) = x^2 - 6")
ax.legend()
ax.grid(True)
plt.show()
````

### 4.3
Plot the absolute error from the true value of $\sqrt{{6}} $.

````python
# Your answer for 4.3 here
import matplotlib.pyplot as plt
import numpy as np
def f(x): return x**2 - 6
def df(x): return 2 * x
x_vals = [2.0]
for _ in range(3):
    x_next = x_vals[-1] - f(x_vals[-1]) / df(x_vals[-1])
    x_vals.append(x_next)

true_value = np.sqrt(6)
errors = [abs(x - true_value) for x in x_vals]
plt.figure(figsize=(10, 6))
plt.plot(errors, marker='o')
plt.xlabel('Iteration')
plt.ylabel('Absolute Error')
plt.title('Absolute Error vs. Iteration')
plt.grid(True)
plt.show()
````

## Part 5-Question 1:

````python
def matrix_multiply(A, B):
    # Your code starts here
    # Hint: Initialize result matrix with 0s, then fill it using 3 nested loops
#def multiply_matrices(A, B):
    m = len(A)
    n = len(A[0])
    n_b = len(B)
    p = len(B[0])

    # Check if matrices can be multiplied
    if n != n_b:
        raise ValueError("Number of columns in A must equal number of rows in B")

    # Allocate result matrix filled with 0s (m x p)
    C = []
    i = 0
    while i < m:
        row = []
        j = 0
        while j < p:
            row.append(0)
            j += 1
        C.append(row)
        i += 1

    # Triple nested loop: C[i][j] = sum_k A[i][k] * B[k][j]
    i = 0
    while i < m:
        j = 0
        while j < p:
            # Compute dot product of A[i][:] and B[:][j]
            total = 0
            k = 0
            while k < n:
                total = total + (A[i][k] * B[k][j])
                k += 1
            C[i][j] = total
            j += 1
        i += 1

    return C
````

# 5.1 How does Newton's Method use tangent to find roots?


# 5.1 Answer:
5.1 How does Newton’s method use tangents to find roots?
Newton’s method finds the roots (zeros) of a function by using the tangent line at an initial guess to approximate where the function crosses the x-axis.
At each step, it replaces the function with its tangent and finds where that tangent intersects the x-axis.
Mathematically:

**x_{n+1} = x_n - f(x)/f'(x)**

This process is repeated until the values converge to a root.

# 5.4 What could go wrong with a poor initial guess in Newton’s method?

A poor initial guess can cause several issues:
•	The method may diverge (move away from the root instead of toward it).
•	It may converge to a different root than expected.
•	The derivative ( f'(x_n) ) could be zero or very small, causing division by zero or unstable results.
•	It may oscillate and fail to converge.
