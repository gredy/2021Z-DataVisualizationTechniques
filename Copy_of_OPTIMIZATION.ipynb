# **Optimization Techniques: Theory, Numerical Examples, and Solver Applications**

## 1. Introduction

Optimization is the mathematical process of selecting the **best decision** from a set of feasible alternatives. It is fundamental in:

* Operations Research
* Data Science
* Engineering Design
* Transportation Systems
* Finance
* Machine Learning

A general optimization problem is written as:

[\min_{x \in \mathbb{R}^n f(x)
subject to
[
g_i(x) \le 0, \quad h_j(x) = 0
]

where:

* ( f(x) ) = objective function
* ( g_i(x) ) = inequality constraints
* ( h_j(x) ) = equality constraints

---

# 2. Major Optimization Techniques

## 2.1 Linear Programming (LP)

### Model Form

[
\min c^T x
]
[
Ax \le b, \quad x \ge 0
]

Characteristics:

* Linear objective
* Linear constraints
* Convex feasible region
* Global optimum guaranteed

---

### Numerical Example (LP)

**Production Planning**

A company produces two products:

| Product | Profit | Machine Hours | Labor Hours |
| ------- | ------ | ------------- | ----------- |
| x₁      | 3      | 2             | 1           |
| x₂      | 5      | 1             | 2           |

Available:

* 100 machine hours
* 80 labor hours

---

### Mathematical Model

[
\max 3x_1 + 5x_2
]

Subject to:

[
2x_1 + x_2 \le 100
]
[
x_1 + 2x_2 \le 80
]
[
x_1, x_2 \ge 0
]

---

### Solution (Analytical)

Solving the intersection:

[
2x_1 + x_2 = 100
]
[
x_1 + 2x_2 = 80
]

Solution:
[
x_1 = 40, \quad x_2 = 20
]

Profit:
[
Z = 3(40) + 5(20) = 120 + 100 = 220
]

---

## 2.2 Integer Programming (IP)

If decision variables must be integers:

[
x \in \mathbb{Z}
]

Applications:

* Facility location
* Scheduling
* Network design

Solution methods:

* Branch and Bound
* Branch and Cut

---

## 2.3 Nonlinear Programming (NLP)

[
\min f(x)
]
where ( f(x) ) is nonlinear.

Example:

[
\min f(x) = x^2 - 4x + 7
]

Derivative:

[
f'(x) = 2x - 4
]

Set equal to zero:

[
2x - 4 = 0
]

[
x = 2
]

Minimum value:

[
f(2) = 4 - 8 + 7 = 3
]

---

## 2.4 Convex Optimization

If:

* ( f(x) ) is convex
* feasible region is convex

Then:

* Any local optimum = global optimum

This is critical in:

* Machine Learning
* Portfolio Optimization
* Regularization problems

# 3.1 Gurobi (gurobipy)

---

### Example Using gurobipy (LP)

```python
import gurobipy as gp
from gurobipy import GRB

# Create model
model = gp.Model("LP_example")

# Variables
x1 = model.addVar(lb=0, name="x1")
x2 = model.addVar(lb=0, name="x2")

# Objective
model.setObjective(3*x1 + 5*x2, GRB.MAXIMIZE)

# Constraints
model.addConstr(2*x1 + x2 <= 100)
model.addConstr(x1 + 2*x2 <= 80)

# Solve
model.optimize()

# Results
print(x1.x, x2.x, model.objVal)
```

Output:

```
x1 = 40
x2 = 20
Objective = 220
```

# 3.2 PuLP (Open Source LP/MIP)

PuLP is Python-based modeling for LP and MIP problems.

---

### Example Using PuLP

```python
from pulp import *

# Create problem
prob = LpProblem("LP_example", LpMaximize)

# Variables
x1 = LpVariable("x1", lowBound=0)
x2 = LpVariable("x2", lowBound=0)

# Objective
prob += 3*x1 + 5*x2

# Constraints
prob += 2*x1 + x2 <= 100
prob += x1 + 2*x2 <= 80

# Solve
prob.solve()

print(value(x1), value(x2), value(prob.objective))
```

---

### Advantages of PuLP

* Open source
* Easy for teaching
* Interfaces with CBC, GLPK, CPLEX, Gurobi

# 4. Real-World Applications

## 4.1 Transportation Optimization

[
\min \sum_{i,j} c_{ij} x_{ij}
]

Applications:

* Urban routing

* Traffic flow optimization

---

## 4.2 Portfolio Optimization

[
\min x^T \Sigma x
]
subject to:
[
\sum x_i = 1
]

Used in:

* Financial engineering
* Risk minimization

---

## 4.3 Machine Learning

Many ML algorithms solve:

[
\min \sum_{i=1}^n (y_i - \hat{y}_i)^2 + \lambda ||w||^2
]

Optimization techniques:

* Gradient descent
* Stochastic methods
* Convex optimization
