## 6. Challenge Problem (Applied Real-World Task)

### Mini Student Gradebook

A teacher wants to compute the class average and highest score from a list of grades.  
Given: `grades = [75, 87, 64, 93, 82, 91, 70, 100, 88, 79]`

**Q16:** Write code to compute and print the average grade (rounded to 2 decimal places).

```python
# Q16: Your code here
grades = [75, 87, 64, 93, 82, 91, 70, 100, 88, 79]
average_grade = sum(grades) / len(grades)
print(round(average_grade, 2))
```

**Q17:** Write code to print the highest grade.

```python
# Q17: Your code here
highest_grade = max(grades)
print(highest_grade)
```

**Q18 (Bonus):** Write a function `grade_report(grades)` that prints both results above and returns nothing. Test your function with the given list.

```python
# Q18: Your code here
def grade_report(grades):
    average_grade = sum(grades) / len(grades)
    highest_grade = max(grades)
    print("Average Grade:", round(average_grade, 2))
    print("Highest Grade:", highest_grade)
    return


grade_report(grades)
```
